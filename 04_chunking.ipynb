**ข้อมูลสมมติสำหรับการอบรม:** ชื่อหน่วยงาน Lot ผู้ขาย ค่าที่วัด และ SOP ในตัวอย่างถูกสร้างขึ้นเพื่อเรียนรู้ AI ไม่ใช่ข้อมูลการผลิตหรือระเบียบจริงของกลุ่มศรีตรัง

บริบทงานอ้างอิงจาก [เว็บไซต์กลุ่มศรีตรัง](https://www.sritranggroup.com/th/home) ส่วนรูปแบบ notebook อ้างอิงจาก [ตัวอย่าง GitHub](https://github.com/biodatlab/cpf-genai-workshop/tree/main/tutorial_notebooks) และใช้ [OpenAI API documentation](https://developers.openai.com/api/docs) สำหรับ syntax

```python
document = '[SOP-1 รับน้ำยาง] บันทึก Lot ผู้ขาย น้ำหนักสุทธิ ทะเบียนรถ และเวลารับเข้า เก็บรหัสผู้ขายเพื่อเชื่อมโยงข้อมูล Traceability\n[SOP-2 QC] เก็บผล DRC พร้อมรหัสตัวอย่างและ Lot หากผลยังไม่ครบ ให้ระบุสถานะรอ QC ห้ามสรุปว่าผ่านเกณฑ์จากข้อมูลที่หายไป\n[SOP-3 คลังสินค้า] Lot ที่รอ QC ต้องแยกในพื้นที่พักรอผลและติดป้าย Lot ตรวจสถานะก่อนยืนยันการจัดส่ง\n[SOP-4 ประสานงาน] ในสถานการณ์ฝึกอบรม ระดับน้ำตั้งแต่ 30 ซม. ให้ประสานหัวหน้าคลังและฝ่ายขนส่งเพื่อตรวจสอบแผนรับส่งสินค้า\n[SOP-5 น้ำท่วม] ในสถานการณ์ฝึกอบรม ระดับน้ำตั้งแต่ 50 ซม. ให้พักแผนโหลดสินค้าของพื้นที่ที่ได้รับผลกระทบ แจ้งหัวหน้าคลัง ฝ่ายขนส่ง และ EHS เพื่อประเมินตามแผนของพื้นที่\n[SOP-6 ส่งมอบ] เมื่อเวลารับส่งเปลี่ยน ให้ฝ่ายขนส่งประสานฝ่ายขาย แจ้งกำหนดการใหม่ที่ยืนยันแล้ว ห้ามรับรองเวลาส่งมอบที่ยังไม่ยืนยัน'
```

```python
def split_text(text, size=150, overlap=30):
    if not 0 <= overlap < size:
        raise ValueError("กำหนด 0 <= overlap < size")
    return [text[start:start + size] for start in range(0, len(text), size - overlap)]

chunks = split_text(document)
for i, chunk in enumerate(chunks, 1):
    print(f"Chunk {i} ({len(chunk)} ตัวอักษร):\n{chunk}\n")
```

## ลองทำด้วยตนเอง
1. ทดลอง size=100 และ overlap=0 จากนั้นเปลี่ยน overlap=30
2. หาคำสั่งที่ถูกตัดตรงขอบ chunk
3. อธิบายว่า overlap ช่วยอย่างไร ตัวอย่างนี้ตัดตามจำนวนตัวอักษร จึงอาจตัดคำหรือสระภาษาไทย
